# Sigma decay curves and top-k edges
Reads runs/study1/results/*.json written by hbwm.probes.evaluate.

```
import json
from pathlib import Path
R = Path('runs/study1/results')
h2 = json.loads((R/'h2.json').read_text())
h4 = json.loads((R/'h4.json').read_text())
```

```
import matplotlib.pyplot as plt
from hbwm.probes.eligibility import BUCKET_NAMES
for stem, r in h2.items():
    plt.plot([r['values'].get(b) for b in BUCKET_NAMES], marker='o', label=stem)
plt.xticks(range(6), BUCKET_NAMES); plt.legend(); plt.ylabel('probe acc'); plt.show()
```

```
for stem, r in h4.items():
    print(stem, 'median k90 =', r['median_k90'], 'strong', r['strong'], 'weak', r['weak'])
```
